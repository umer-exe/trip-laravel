# Atlas AI Chatbot Testing Notebook

This notebook allows you to interactively test and fine-tune the chatbot from `app.py`.
- Automatically handles concise vs detailed responses.
- Detects irrelevant questions and responds appropriately.
- Stores conversation history for review.

```python
# --- Import chatbot from app.py ---
from app import conversation_chain

# --- Initialize chat log ---
chat_log = []

# --- Helper function for asking questions ---
def ask(question: str):
    """Ask a question to the chatbot and print/store the response."""
    if not question.strip():
        print("Question cannot be empty.")
        return

    # Check for detail keywords
    detail_keywords = ['detail', 'more info', 'full description', 'explain', 'elaborate']
    if any(word.lower() in question.lower() for word in detail_keywords):
        question_to_bot = question + " Provide full detailed information about the tours mentioned."
    else:
        question_to_bot = question + " Keep the answer concise and clear."

    # Detect irrelevant questions based on context keywords (simple heuristic)
    allowed_keywords = ['tour', 'package', 'travel', 'location', 'price', 'duration', 'highlights', 'overview']
    if not any(word.lower() in question.lower() for word in allowed_keywords):
        response = "Sorry, I can only answer questions about tours and travel packages."
    else:
        try:
            result = conversation_chain({"question": question_to_bot})
            response = result['answer']
        except Exception as e:
            response = f"[ERROR] Chatbot failed: {e}"

    # Print and store
    print(f"Q: {question}")
    print(f"A: {response}\n")
    chat_log.append({'question': question, 'answer': response})
    return response
```

## Example usage
Type your question in the `ask()` function and run the cell to see chatbot responses.

```python
# Example questions
ask("What are the top tours?")
ask("Give me detailed information about the x package.")
ask("Tell me about sports in Karachi")  # Should trigger irrelevant response
```
